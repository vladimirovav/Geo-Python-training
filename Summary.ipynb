```python
def celsius_to_fahr(temp_celsius):
    """Converts temperatures from Celsius to Fahrenheit.

    Parameters
    ----------

    temp_celsius: int | float
        Input temperature in Celsius (should be a number)

    Returns
    -------

    Temperature in Fahrenheit (float)
    """

    # Convert the Celsius into Fahrenheit
    converted_temp = (temp_celsius * 1.8) + 32

    return converted_temp
```

```python
fp = r"data/Rovaniemi_railway_station.csv"
data = pd.read_csv(
    fp,
    usecols=['Year', 'Month', 'Day','Time [Local time]', 'Air temperature mean [°C]'])

# New column names
new_names = {"Observation station": "STATION", "Time [Local time]": "TIME", "Air temperature mean [°C]": "TEMP_C"}

# Rename new columns:
data = data.rename(columns=new_names)

# Turn column into numeric:
data["TEMP_C"] = pd.to_numeric(data["TEMP_C"], errors="coerce")
data["TEMP_F"] = data["TEMP_C"].apply(celsius_to_fahr)

# Convert to string (because it is easier to work with a string when creating a new column)
data["MONTH_STR"] = data["Month"].astype(str)
data["DAY_STR"] = data["Day"].astype(str)
data["MONTH_DAY"] = data["MONTH_STR"].str.zfill(2) + "-" + data["DAY_STR"].str.zfill(2)

# Group by month and day
grouped = data.groupby("MONTH_DAY")

# Specify the columns that will be part of the calculation
mean_cols = ["TEMP_C", "TEMP_F"]

# Create an empty DataFrame for the aggregated values
daily_data = pd.DataFrame()

# Iterate over the groups
for key, group in grouped:

    # Calculate mean for each group
    mean_values = group[mean_cols].mean()

    # Add the ´key´ (i.e. the date+time information) into the aggregated values
    mean_values["MONTH_DAY"] = key
    
    # Convert the mean_values series to a DataFrame and make it have a row orientation
    # to frame will convert the data, BUT it will also write the values vertically
    # SO, you need to transpose it, which will swap rows for columns
    row = mean_values.to_frame().transpose()

    # Concatenate the aggregated values into the daily_data DataFrame
    # concat - append it to the list:
    daily_data = pd.concat([daily_data, row], ignore_index=True)

print(daily_data)
```

```python
data = pd.read_csv("data/2315676.txt",
                   # criteria for what becomes an NA:
                   na_values=["-","*","***","-9999"], 
                   # Separation with white space:
                   sep=r"\s+", 
                   # Skip second row (i.e., 1)
                   skiprows=[1],
                   # note that the header is present:
                   header = 0)
```

```python
data.head(15)
data.tail(15)
```

```python
# Fill NAs with a calculation:
data["TAVG"] = data["TAVG"].fillna((data["TMAX"] + data["TMIN"]) / 2)
data.tail(15)
```

```python
# Drop TAVG values that have missing NA
data_clean = data.dropna(subset = ["TAVG"])
len(data_clean)
```

```python
len(data)
```

```python
data["TAVG_fahr"] = celsius_to_fahr(data["TAVG"])
data.head()
```

**Seasons**:

Winter: December, January, February

Spring: March, April, May

Summer: June, July, August

Autumn: September, October, November

```python
# To create a seasons column
# 1. change the date to datetime format:
data["DATE"] = pd.to_datetime(data["DATE"], format="%Y%m%d")
data.dtypes
data.head()
```

```python
# 2. Create a months column:
# Note if else is called Boolean values (where there are true or false statements)
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"

# function from dt (i.e., date) called month (so essentially it grabs the month from the date column
# apply - tells to run the function (get_season) to every single variable
data["Season"] = data["DATE"].dt.month.apply(get_season)
data.head()
```

```python
# Calculate seasonal averages for the reference period 1951-1980
Reference_data = data[
    (data["DATE"].dt.year <= 1980) & (data["DATE"].dt.year >= 1951)
    ]

seasonal_means = (
    Reference_data
    .groupby("Season")["TAVG_fahr"]
    .mean()
    # Note, here you want seasons to be an index because .map function works with arrays, not dataframes
    #.reset_index()
)
seasonal_means
```

```python
# Compute seasonal mean for each year
# Create a year column
data["Year"] = data["DATE"].dt.year

# create seasonal averages per year:
seasonal_year = (
    data
    # group by Year-Season
    # and select the column TAVG to manipulate further:
    .groupby(["Year", "Season"])["TAVG_fahr"]
    # calculate mean:
    .mean()
    # recreate index:
    .reset_index()
)
seasonal_year
```

```python
# Merge the reference TAVG averages with the normal one 
# Create on what you want to reference the seasons
data["TAVG_ref"] = data["Season"].map(seasonal_means)

# Since there are no negative values in TAVG_ref (since it is in Fahrengheit, we can calcualte simply)
data["TAVG_anom"] = data["TAVG_fahr"] - data["TAVG_ref"]
```

```python
# Now only select data from 1980s
data_new = data[data["DATE"].dt.year > 1980]
data_new.head()
```

```python
# Create new data.frames with seasons:
Winter = data_new[data_new["Season"]== "Winter"]
Summer = data_new[data_new["Season"]== "Summer"]
Spring = data_new[data_new["Season"]== "Spring"]
Autumn = data_new[data_new["Season"]== "Autumn"]
# Other signs include <= less than, >= more than, != Not equal
```

```python
#### PLOT ########
# Plot
Winter.plot(x="DATE", 
            y="TAVG_anom", 
            # kind of plot: line
            kind="line", 
            #figure size in inches
            figsize=(12, 4),
           xlabel = "Date",
           ylabel = "Temperature anomaly (F°)",
           title = "Winter Temperature Anomalies Over Time",
           grid = True)
```

```python
# Since you have rows and columns, it is nested
fig, axs = plt.subplots(nrows=2, ncols=2, figsize=(12, 8))
axs

# Rename your axis (of the plots) to be able to access them easier:
ax11 = axs[0][0]
ax12 = axs[0][1]
ax21 = axs[1][0]
ax22 = axs[1][1]

# Plot data
# plot to which axis you want to add the visualisation to, colour of the line
Winter.plot(ax=ax11, x="DATE", y="TAVG_anom",c="blue")
Spring.plot(ax=ax12, x="DATE", y="TAVG_anom",c="orange")
Summer.plot(ax=ax21, x="DATE", y="TAVG_anom",c="green")
Autumn.plot(ax=ax22, x="DATE", y="TAVG_anom",c="brown")

# Display the plot
# Note: This is not required, but suppresses text from being printed
# in the output cell
plt.show()
```
