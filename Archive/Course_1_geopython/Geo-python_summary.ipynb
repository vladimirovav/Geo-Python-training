# POINTS

```python
# Import 'shapely.geometry.point' class
from shapely import Point

# Create 'Point' objects:
point1 = Point(2.2, 4.2) #  x and y coordinates
point2 = Point(7.2,-25.1)
point3 = Point(9.26,-2.456)
point4_3D = Point(9.26,-2.456,0.57)
```

```python
# if you use print you will get the coordinates
print(point1)
print(point4_3D)
type(point1)
point1.geom_type
```

```python
# Get coordinate tuple(s)
list(point1.coords)

# to get the coordinates:
x = point1.x
y = point1.y
print(x,y)
```

```python
# Calculate distance:
dist = point1.distance(point2)
print(f"Distance between the points is {dist:.2f} units") # 2f rounds up to 2 decimals
```

# LINES

```python
from shapely.geometry import LineString

# Create a line:
line = LineString([point1, point2, point3])

# Create a LineString from a list of coordinates:
line2 = LineString([(2.2, 4.2),(7.2,-25.1),(9.26,-2.456)])

# Check if the lines are identical:
line == line2

line
# type(line) # should be shapely.geometry.linestring.LineString

# Obtain x and y coordinates of a line:
xcoords = list(line.xy[0])
ycoords = list(line.xy[1])

print(xcoords)
print(ycoords)

# Calculate the length of the line:
line.length

# Calculate the centre of the line:
print(line.centroid)
```

# Polygons

```python
from shapely.geometry import Polygon
from shapely.geometry import LinearRing

# Creat a Polygon from the coordinates
polygon1 = Polygon([(2.2, 4.2),(7.2,-25.1),(9.26,-2.456)])
polygon2 = Polygon([point1, point2, point3])

# Create LinearRing
#LinearRing - the begining and the end of the points will be the same
shell = LinearRing([point1, point2, point3])

shell
```

```python
#define the exterior:
outer = LinearRing([(-180,90), (-180,-90),(180,-90),(180,90)])
#define a hole:
hole = LinearRing([(-170,80),(-100,-80),(100,-80),(170,80)])

polygon_with_hole = Polygon(outer, [hole])
polygon_with_hole
polygon_with_hole.exterior.length
polygon_with_hole.exterior
polygon_with_hole.bounds
polygon_with_hole.area
polygon_with_hole.interiors
```
