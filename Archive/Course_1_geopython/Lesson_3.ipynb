```python
yesterday = 14
today = 10
tomorrow = 13
if yesterday <= today: # statement is wrong
    print("A")
elif today != tomorrow: # first correct statement will  be printed
    print("B")
elif yesterday > tomorrow:
    print("C")
elif today == today:
    print ("D")
    
```

**Combine for loops and conditional statements**

```python
temperatures = [0,28,12,17,30]
# this will go through every single variable in temperatures:
for temperature in temperatures:
    if temperature > 25:
        print(f"{temperature} it is hot")
    else:
        print(f"{temperature} it is not hot")
    
```
